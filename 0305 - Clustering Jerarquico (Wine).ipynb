````python
## Librerias
from ucimlrepo import fetch_ucirepo 
import matplotlib.pyplot as plt
import seaborn as sns
from pandas import DataFrame

from sklearn.preprocessing import StandardScaler
from scipy.cluster.hierarchy import linkage, dendrogram, fcluster
from sklearn.decomposition import PCA
````

## Detalles de funciones

Utilizaremos las funciones linkage, dendrogram y fcluster de la biblioteca `scipy` que se encuentra en la sección `cluster.hierarchy`<br>

La función `linkage` aprende el modelo definido y retorna un dendrograma<br>
```{python}
    linkage(y, method='single')
```

| Parámetros | Descripción |
|------------|-------------|
| y      | Representa los datos usado para el modelo |
| method | Representa los linkage como: 'ward', 'complete', 'average', 'single' |


Para graficar se usa la función `dendrogram` con el modelo aprendido
```{python}
    dendrogram(model, p=30, truncate_mode=None)
```

| Parámetros | Descripción |
|------------|-------------|
| model         | Representa el modelo ajustado |
| p             | Párametro usado para truncate_mode |
| truncate_mode | None (muestra todo el dendrograma), 'lastp' (corta el dendrograma después de p ramas), 'level' (muestra un máximo de p niveles) |

Para extraer los clusters se utiliza la función `fcluster`
```{python}
    fcluster(Z, t, criterion='distance')
```

| Parámetros | Descripción |
|------------|-------------|
| Z | es el modelo generado por la función linkage |
| t | es el valor donde se quiere cortar el dendrograma para generar los cluteres |
| criterion | el criterio de corte. 'distance' es el que nos permite generar los clusters, basados en la altura que definimos.|


## Carga de datos

__Dataset:__

Estos datos son referenciados en [UCI Machine Learning Repository](https://archive.ics.uci.edu/dataset/109/wine) que son los resultados de un análisis químico de vinos cultivados en una misma región de Italia pero derivados de cultivos diferentes.

````python
## Cargar objeto de datos
dataset = fetch_ucirepo(id=109)

## Extraer los features del dataset
data = dataset.data.features 

## Mostrar los primeros 5 registros
display(data.head())

## Escalado de los datos
## Instancia del modelo de escalado
scaler = StandardScaler()

## Ajuste del modelo y transformación de los datos
data_scaled = scaler.fit_transform(data)
print('(shape) data: {}'.format(data_scaled.shape))
````

````python
#Entrenando el modelo
model = linkage(data_scaled, method='complete')

## Visualización del dendrograma
plt.figure(figsize=(15, 7)) 
object_dendrog = dendrogram(Z=model, p=30, truncate_mode='none')
````

````python
## Extraer los clusters sugeridos por el método
clusters=fcluster(model, t=8.0, criterion="distance")
clusters
````

````python
## Transformación de los datos usando PCA
pca_model = PCA(n_components=2)
pca_model.fit(data_scaled)
pca_data = pca_model.transform(data_scaled)

## Crear un dataframe con los datos transformados mediante PCA y agregar las etiquetas de los clusteres
pca_data = DataFrame(pca_data, columns=['PC1', 'PC2'])
pca_data['cluster'] = clusters
pca_data
````

````python
## Graficar los datos por clases. 
N = pca_data['cluster'].nunique()

plt.figure(figsize=(8, 6))
sns.scatterplot(data=pca_data, 
                x='PC1', y='PC2', 
                hue='cluster',
                palette=sns.color_palette()[:N])
plt.legend(loc=[1.01, 0.5], title='Cluster')
plt.tight_layout()
plt.show()
````
